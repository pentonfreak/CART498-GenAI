```python
import os
import re
import sys
from openai import OpenAI

# Lift the limit for integer-to-string conversion
sys.set_int_max_str_digits(0)

try:
    from google.colab import userdata
    api_key = userdata.get('OpenAI-Token')
except Exception:
    api_key = os.environ.get("OpenAI-Token") or os.environ.get("OpenAI-Token")

# Enter your API key in Colab Secrets or directly here for testing
client = OpenAI(api_key=api_key)

MODEL = "gpt-4.1-nano"


def ask_gpt(number):
    """
    Ask GPT to calculate number * number.
    The bot is intentionally asked to provide only the numerical answer.
    """

    prompt = f"""
Calculate the following multiplication:

{number} * {number}

Return ONLY the numerical answer.
Do not explain your answer.
"""

    response = client.responses.create(
        model=MODEL,
        input=prompt
    )

    answer = response.output_text.strip()

    # Try to extract an integer from GPT's response
    match = re.search(r"-?\d+", answer)

    if match:
        return int(match.group())

    return None


def self_deprecating_message(error_count):
    """
    Generate progressively more frustrated responses
    as the bot makes more mistakes.
    """

    messages = [
        "Oops. I somehow messed up basic multiplication.",
        "Wow, I got that wrong again. Math is not going well for me.",
        "Okay... this is getting embarrassing. I really need to study multiplication.",
        "Another mistake?! At this point, I should probably not be trusted with a calculator.",
        "I have officially lost an argument with elementary-school mathematics.",
        "That's another one. I'm beginning to question my entire existence as a math bot."
    ]

    index = min(error_count - 1, len(messages) - 1)
    return messages[index]


def run_math_bot(n, iterations):
    """
    Run the repeated self-multiplication experiment.
    """

    current_number = n
    error_count = 0

    print(f"Starting with n = {n}")
    print(f"Iterations = {iterations}")
    print("-" * 50)

    for i in range(1, iterations + 1):

        # Correct mathematical calculation
        correct_result = current_number * current_number

        # GPT calculation
        ai_result = ask_gpt(current_number)

        print(f"\nIteration {i}")
        print(f"{current_number} × {current_number}")

        print(f"Correct answer: {correct_result}")
        print(f"AI answer:      {ai_result}")

        if ai_result == correct_result:
            print("✓ Correct!")

        else:
            error_count += 1

            print("✗ INCORRECT!")
            print(self_deprecating_message(error_count))

        # The next iteration uses the CORRECT result.
        current_number = correct_result

    print("\n" + "-" * 50)
    print(f"Experiment finished.")
    print(f"Total AI errors: {error_count}")

    if error_count == 0:
        print("Unfortunately, I survived the experiment without embarrassing myself.")
    else:
        print(
            f"I made {error_count} mistake(s). "
            "Please never ask me to do your taxes."
        )

    return error_count


n = int(input("Enter the base number n: "))
iterations = int(input("Enter the number of iterations i: "))

run_math_bot(n, iterations)
```
